```r
library(tidyverse)
library(dplyr)
library(readr)
```

```r
#load population
lsoa_pop <-read_csv(
  "/content/LSOA_population.csv",
  skip=6
  )
lsoa_pop <- lsoa_pop[-1,]
lsoa_pop <- lsoa_pop[1:(nrow(lsoa_pop)-5), ]
```

```r
#load tenure
lsoa_tenure <- read_csv(
  "/content/LSOA_Tenure_London.csv",
  skip = 6
)
lsoa_tenure <- lsoa_tenure[-1,]
lsoa_tenure <- lsoa_tenure[1:(nrow(lsoa_tenure)-5), ]
head(lsoa_tenure)
```

```r
#load disability
lsoa_disability <- read_csv(
  "/content/LSOA_Disability_London.csv",
  skip=6
)
lsoa_disability <- lsoa_disability[1:(nrow(lsoa_disability)-3), ]
```

```r
#load health
lsoa_health <- read_csv(
  "/content/LSOA_Health_London.csv",
  skip = 6
)
lsoa_health <- lsoa_health[-1,]
lsoa_health <- lsoa_health[1:(nrow(lsoa_health)-5), ]
```

```r
#load Ns-Sec
lsoa_Ns_SeC <- read_csv(
  "/content/LSOA_NS_SeC_London.csv",
  skip=6
)
lsoa_Ns_SeC <- lsoa_Ns_SeC[1:(nrow(lsoa_Ns_SeC)-3), ]
```

```r
#load hours worked
lsoa_hours_worked <- read_csv(
  "/content/LSOA_Hours_Worked_London.csv",
  skip=6
  )
lsoa_hours_worked <- lsoa_hours_worked[1:(nrow(lsoa_hours_worked)-3), ]
nrow(lsoa_hours_worked)
```

```r
#economic activity
lsoa_econ_act <- read_csv("/content/LSOA_Economic_Activity_London.csv",
                          skip = 6

)
lsoa_econ_act <- lsoa_econ_act[-1,]
lsoa_econ_act <- lsoa_econ_act[1:(nrow(lsoa_econ_act)-5), ]
nrow(lsoa_econ_act)
```

```r
#data qualifcation
lsoa_qaulifcation <- read_csv(
  "/content/LSOA_Qualification_London.csv",
  skip=6
)
lsoa_qaulifcation <- lsoa_qaulifcation[-1,]
lsoa_qaulifcation <- lsoa_qaulifcation[1:(nrow(lsoa_qaulifcation)-5), ]
nrow(lsoa_qaulifcation)
```

```r
#combine
lsoa_all <- lsoa_pop %>%
  left_join(lsoa_tenure, by = c("2001 super output areas - lower layer","mnemonic","row")) %>%
  left_join(lsoa_disability, by = c("2001 super output areas - lower layer","mnemonic","row")) %>%
  left_join(lsoa_health, by = c("2001 super output areas - lower layer","mnemonic","row")) %>%
  left_join(lsoa_Ns_SeC, by = c("2001 super output areas - lower layer","mnemonic","row")) %>%
  left_join(lsoa_hours_worked, by = c("2001 super output areas - lower layer","mnemonic","row")) %>%
  left_join(lsoa_econ_act, by = c("2001 super output areas - lower layer","mnemonic","row")) %>%
  left_join(lsoa_qaulifcation, by = c("2001 super output areas - lower layer","mnemonic","row"))

```

```r
lsoa_all <- lsoa_all %>%
  mutate(
    borough_name = sub(" \\d{3}[A-Z]$", "", `2001 super output areas - lower layer`)
  )

  ncol(lsoa_all)


  inner_lodnon <- c("Camden", "Hackney", "Hammersmith and Fulham", "Haringey", "Islington",
  "Kensington and Chelsea", "Lambeth", "Lewisham", "Newham",
  "Southwark", "Tower Hamlets", "Wandsworth", "Westminster")

  lsoa_all$inner_london_T <- lsoa_all$borough_name %in% inner_lodnon


```

```r
#export

write.csv(lsoa_all, "lsoa_all.csv")
```
